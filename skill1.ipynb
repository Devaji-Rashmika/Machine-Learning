```python
N = int(input("Enter a number:"))
RN = 0
while N>0:
    R = N%10
    RN =( RN * 10) + R
    N = N//10
print(RN)
```

```python
def isPrime(n):
    if n < 2:
        print("Not Prime")
    elif n == 2:
        print("Prime")
    else:
        for i in range(2, int(n**0.5) + 1):
            if n % i == 0:
                print("Not Prime")
                break
        else:
            print("Prime")

n = int(input("Enter number: "))
isPrime(n)
```
